```python
import numpy as np
import pandas as pd
```

```python
def load_santa_fe(path):
    """
    Load the Santa Fe competition datasets.
    """
    
    with open(path) as f:
        text = f.read()
    values = [v for v in text.replace("\n", ",").split(",") if v.strip()]
    return pd.DataFrame(values, columns=["intensity"])
```
